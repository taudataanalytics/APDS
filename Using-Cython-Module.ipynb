## Native Python function

```python
def foo(check):
    
    count = 0
    
    for i in range(10000):
        for j in range(10000):
            if (i+j)%11 == check:
                count += 1
                
    return count
```

```python
%timeit foo(2)
```

```python
foo(2)
```

## Modify function

```python
%%cython -a

def foo_c(int check):
    
    cdef int i,j
    cdef int count = 0
    
    for i in range(10000):
        for j in range(10000):
            if (i+j)%11 == check:
                count += 1
                
    return count
```

```python
%timeit foo_c(2)
```

```python
foo_c(2)
```
