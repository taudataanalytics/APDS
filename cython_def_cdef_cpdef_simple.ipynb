## Quick conceptual summary
- **`def`** — normal Python function. Easy to write and debug; callable from Python.
- **`cdef`** — C-level function (and C-level variables). Very fast when called from Cython/C, but **not** directly callable from Python.
- **`cpdef`** — hybrid: provides a C-level function for fast internal calls and also exposes a Python wrapper so Python code can call it.

## Simple analogy
- `def` = the public entrance (Python users).
- `cdef` = staff-only entrance (Cython internal calls).
- `cpdef` = both entrances side-by-side (fast internal access + Python access).

## Tiny examples — all do the same operation (add two numbers)

These three examples implement `add(a, b)` in three different ways so beginners can focus on callability and performance implications, not algorithmic complexity.

### Example 1 — `def` (Python-level)
A normal Python-callable function.

````python
%%cython
# Example: def (Python-callable)
def add_def(int a, int b):
    """Python-callable function. Good default for most use-cases."""
    return a + b

print('add_def(2, 3) =', add_def(2, 3))
````

### Example 2 — `cdef` (C-level only)
`cdef` defines a C function. It is fast for internal calls but not available directly from Python. We add a small `def` wrapper so you can call it from Python for demonstration.

````python
%%cython
# Example: cdef (C-only) + Python wrapper
cdef int add_cdef(int a, int b):
    """C-level function — not exported to Python."""
    return a + b

# Python wrapper to call the cdef from Python
def call_add_cdef(int a, int b):
    return add_cdef(a, b)

print('call_add_cdef(2, 3) =', call_add_cdef(2, 3))
````

### Example 3 — `cpdef` (hybrid)
`cpdef` creates a C implementation plus a Python wrapper automatically. Use it when you want both Python accessibility and fast internal calls.

````python
%%cython
# Example: cpdef (both C-level and Python wrapper)
cpdef int add_cpdef(int a, int b):
    """Hybrid: callable from Python and fast for internal C-level calls."""
    return a + b

print('add_cpdef(2, 3) =', add_cpdef(2, 3))
````

## Demonstration: Python call vs internal (conceptual)

- When you call `add_def` from Python, it uses Python calling machinery.
- When you call `add_cpdef` from Python, Python calls the wrapper which converts arguments and calls the C implementation.
- When Cython code calls the `cpdef` C function directly, it bypasses the wrapper and is faster.

The examples above show that all three return the same result; the difference is in visibility and call overhead.

## Comparison table

| Declaration | Callable from Python? | Callable from C/Cython? | Typical use |
|---|---:|---:|---|
| `def` | ✔ Yes | ✖ No | Regular Python-facing functions |
| `cdef` | ✖ No | ✔ Yes | Internal fast helpers (hot code paths) |
| `cpdef` | ✔ Yes (wrapper) | ✔ Yes (direct C call) | When you want both Python API and fast internal calls |

## Recommendations for beginners
- Start with `def`. It is simple and clear.
- If profiling shows a hotspot that is only called from Cython, convert it to `cdef`.
- If you need both Python access and fast internal calls, use `cpdef`.
- Keep examples small and prefer clear wrappers when you need to test `cdef` functions from Python.
