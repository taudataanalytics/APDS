<center><h2><strong><font color="blue"> Introduction to Functional Programming </font></strong></h2></center>

Functional Programming (FP) is a declarative programming paradigm that treats computation as the evaluation of mathematical functions and avoids changing-state and mutable data. In Data Science, FP is particularly valuable for building robust data pipelines, ensuring reproducibility, and facilitating parallel processing.

**Core Principles**

1.	**Immutability**: Data cannot be modified after creation. Instead of changing a variable, you create a new one with the updated value. This reduces bugs related to unintended side effects.
2.	**Pure Functions**: A function's output relies solely on its input arguments and produces no side effects (e.g., modifying global variables or printing to the console). Given the same input, a pure function always returns the same output.
3.	**First-Class Functions**: Functions are treated as citizens of the language. They can be assigned to variables, passed as arguments, and returned from other functions.

<center><img alt="" src="images/APDS/intro-to-FP.jpg" style="height: 360px;"/></center>

## Immutability

Immutability refers to the principle that once a piece of data is created, it cannot be altered; instead, any “change” is represented by creating a new data object with the desired modification. This approach significantly reduces bugs caused by unintended side effects, as functions cannot accidentally modify shared data that is used elsewhere in a program. In data analysis and scientific computing, immutability enhances reliability and reproducibility because the original data remains intact throughout the computational process, making program behavior easier to reason about and verify.

````python
# Immutable approach: creating new data instead of modifying existing data

original_data = (1, 2, 3)          # a tuple is immutable
updated_data = original_data + (4,)  # create a new tuple with an added value

print("Original data:", original_data)
print("Updated data:", updated_data)
````

## Pure Functions

Pure functions are functions whose outputs depend exclusively on their input arguments and that do not produce side effects, such as modifying external variables, altering data structures in place, or interacting with input/output operations. This property guarantees referential transparency: for a given input, a pure function will always return the same output, regardless of when or how often it is executed. In practice, pure functions improve code reliability, simplify debugging and testing, and support parallel and distributed computation, since their execution does not interfere with shared state.

````python
# Example of a pure function

def square(x):
    return x * x

result1 = square(4)
result2 = square(4)

print(result1)  # 16
print(result2)  # 16
````

# First-class functions

First-class functions refer to the language property whereby functions are treated as ordinary values, meaning they can be assigned to variables, passed as arguments to other functions, and returned as results of function calls. 

In Python, this capability enables higher-order programming patterns such as callbacks, function composition, and strategy-based design, all of which are central to functional programming and modern software engineering. 

By allowing behavior to be abstracted and manipulated independently of data, first-class functions promote modularity, reuse, and expressive code structures, particularly in data processing pipelines where transformation logic can be dynamically composed and reused. This paradigm also underpins many built-in functional constructs in Python, including `map`, `filter`, decorators, and lambda expressions, making first-class functions a foundational concept for writing flexible, declarative, and maintainable programs.

<center><img alt="" src="images/APDS/first-class-function.jpg" style="height: 360px;"/></center>

````python
# A simple function
def greet(name):
    return f"Hello, {name}"

# 1. Assigning a function to a variable
say_hello = greet

# 2. Passing a function as an argument
def apply_function(func, value):
    return func(value)

# 3. Returning a function from another function
def get_greeting_function():
    return greet

# Using the function in all three ways
result1 = say_hello("Alice")
result2 = apply_function(greet, "Bob")
greeting_func = get_greeting_function()
result3 = greeting_func("Charlie")

print(result1)   # Hello, Alice
print(result2)   # Hello, Bob
print(result3)   # Hello, Charlie
````

<center><h2><strong><font color="blue"> Fundamental Concepts and Syntax </font></strong></h2></center>

<center><img alt="" src="images/APDS/pure-lambda-FP.jpg" style="height: 480px;"/></center>

In Functional Programming for Data Science, **pure functions** are preferred because they depend solely on their input parameters and produce no side effects, ensuring that data transformation steps are predictable, testable, and reproducible. 

In contrast, **impure functions** rely on or modify external state, such as global variables, which introduces hidden dependencies and can lead to non-deterministic behavior. 

Complementing this principle, Python’s **lambda functions** provide a concise mechanism for defining small, anonymous functions intended for simple, often one-time operations. Although lambda functions are not inherently pure, they are frequently used to express pure logic within functional constructs such as `map` and `filter`, thereby supporting a declarative and function-oriented programming style.

````python
# Example Impure Function (Avoid in FP):
total = 0

def add_to_total(amount):
    """Relies on and modifies global state."""
    global total
    total += amount
    return total
````

````python
# Example Pure Function (Preferred):
def add(a, b):
    """Depends only on inputs; no side effects."""
    return a + b
````

Lambda functions are not inherently pure because functional purity depends on a function’s behavior rather than on the syntactic form used to define it. In Python, a lambda expression can freely capture and depend on variables from its surrounding scope, including mutable or global state, meaning its output may change even when its explicit input arguments remain the same. As a result, a lambda function may violate the core requirement of purity—that identical inputs must always produce identical outputs—if it relies on external state or causes side effects. For example, the lambda function below is impure because its result depends on the external variable `offset`, which can change independently of the function’s input:

```python
offset = 10
add_offset = lambda x: x + offset

add_offset(5)  # Returns 15

offset = 20
add_offset(5)  # Returns 25 (same input, different output)


# Lambda functions

Lambda functions in Python provide a concise mechanism for defining small, anonymous functions intended for simple, short-lived computations. Introduced using the `lambda` keyword, a lambda function consists of a list of arguments followed by a single expression, whose value is implicitly returned. 

This compact syntax makes lambda functions particularly suitable for one-time operations, such as inline transformations within functional constructs like `map`, `filter`, or `sorted`, where defining a full named function would be unnecessarily verbose. 

Although a standard function definition and its lambda equivalent may express identical logic—as in the computation of a square—lambda functions emphasize brevity and expressiveness, reinforcing a declarative programming style while remaining subject to the same semantic rules as regular Python functions.

<center><img alt="" src="images/APDS/lambda-func-python.jpg" style="height: 480px;"/></center>

````python
# Standard function
def square(x):
    return x ** 2

# Equivalent Lambda function
square_lambda = lambda x: x ** 2

print(square_lambda(5)) # Output: 25
````

<center><h2><strong><font color="blue"> Higher-Order Functions </font></strong></h2></center>

Higher-order functions are a fundamental element of Functional Programming in Python, as they allow functions to be passed as arguments or returned as values, enabling a declarative and expressive approach to data processing. The built-in functions `map`, `filter`, and `reduce` exemplify this paradigm by abstracting common patterns of iteration over collections: `map()` applies a transformation to each element, `filter()` selects elements based on a predicate, and `reduce()` aggregates values through a cumulative operation. Together, these functions support clear, concise, and reusable data transformation pipelines, which are particularly valuable in Data Science tasks such as data cleaning, transformation, and aggregation.

<center><img alt="" src="images/APDS/higher-order-functions-FP-python.jpg" style="height: 360px;"/></center>

## Map

The map() function applies a specific function to every item in an iterable (like a list) and returns an iterator.

Use Case: Transforming a column of data (e.g., unit conversion).

````python
data = [1, 2, 3, 4, 5]

# objective: Multiply every number by 10
# Syntax: map(function, iterable)
result = list(map(lambda x: x * 10, data))

print(result) 
````

## Filter
The filter() function constructs an iterator from elements of an iterable for which a function returns true.

Use Case: Cleaning data by removing nulls or outliers.

````python
data = [10, -5, 20, -1, 50]

# Objective: Keep only positive numbers
filtered_data = list(filter(lambda x: x > 0, data))

print(filtered_data) 
````

## Reduce
The reduce() function, found in the functools module, applies a rolling computation to sequential pairs of values in a list.

Use Case: Aggregation (Sum, Product, Max, Min).

````python
from functools import reduce

data = [1, 2, 3, 4]

# Objective: Calculate the product of all elements
# Logic: ((1 * 2) * 3) * 4
product = reduce(lambda x, y: x * y, data)

print(product) 
````

<center><h2><strong><font color="blue"> List Comprehensions: The Pythonic Alternative </font></strong></h2></center>

List comprehensions in Python offer a clear and expressive alternative to traditional functional programming tools such as `map()` and `filter()`, achieving the same goals of data transformation and selection in a more readable and Pythonic manner. By combining iteration, transformation, and conditional logic within a single construct, list comprehensions make code easier to understand and maintain while preserving a functional, side-effect-free style. This approach is particularly effective in data analysis and scientific computing, where concise and transparent data preprocessing and feature transformation are essential for correctness and reproducibility.

<center><img alt="" src="images/APDS/python-list-comprehension.jpg" style="height: 360px;"/></center>

````python
data = [1, 2, 3, 4, 5]

# Map equivalent
squared = [x ** 2 for x in data]

# Filter equivalent
evens = [x for x in data if x % 2 == 0]

# Map and Filter combined
even_squares = [x ** 2 for x in data if x % 2 == 0]

even_squares
````

````python
# Traditional (Imperative) Version
numbers1 = [1, 2, 3]
numbers2 = [4, 5, 6]

even_sum_pairs = []

for x in numbers1:
    for y in numbers2:
        if (x + y) % 2 == 0:
            even_sum_pairs.append((x, y))

print(even_sum_pairs)
````

````python
# List Comprehension version

[(x, y) for x in numbers1 for y in numbers2 if (x + y) % 2 == 0]
````

<center><h2><strong><font color="blue"> Case Study: A simple E-Commerce Transaction Pipeline </font></strong></h2></center>

This case study demonstrates how a functional programming pipeline can be effectively applied to an e-commerce transaction dataset to clean, normalize, transform, and aggregate data in a clear and reliable manner. By using `filter` to remove invalid records, `map` with a pure function to standardize user names and convert currencies, and `reduce` to calculate total revenue, the solution emphasizes modularity, immutability, and declarative logic. The resulting linear workflow improves readability, reproducibility, and maintainability, making it well suited for real-world data processing and analytical tasks.

<center><img alt="" src="images/APDS/caseStudy-FP-ecommerce.jpg" style="height: 360px;"/></center>

## Scenario: 
You possess a list of raw transaction records (dictionaries). The dataset contains valid and invalid entries, mixed currencies, and varying casing in names.

## Objective: 
1. Clean: Remove transactions with missing amounts.
2. Normalize: Convert user names to title case.
3. Transform: Convert USD amounts to EUR (Exchange rate: 0.85).
4. Aggregate: Calculate the total revenue in EUR.

### Raw Data:

````python
transactions = [
    {'user': 'ALICE', 'amount': 100, 'currency': 'USD'},
    {'user': 'bob', 'amount': None, 'currency': 'USD'},  # Invalid
    {'user': 'CHARLIE', 'amount': 50, 'currency': 'USD'},
    {'user': 'dave', 'amount': 200, 'currency': 'USD'}
]
transactions
````

## The Functional Solution

````python
from functools import reduce

# Constants
EXCHANGE_RATE = 0.85

# Step 1: Filter (Clean Data)
# Keep transactions where amount is not None
valid_transactions = list(filter(lambda t: t['amount'] is not None, transactions))
valid_transactions
````

````python
# Step 2: Map (Transformation Pipeline)
# We define a pure function to handle row transformation
def process_record(record):
    return {
        'user': record['user'].title(),
        'amount_eur': record['amount'] * EXCHANGE_RATE
    }

processed_data = list(map(process_record, valid_transactions))
processed_data
````

````python
# Step 3: Reduce (Aggregation)
# Sum the 'amount_eur' fields
total_revenue = reduce(lambda acc, curr: acc + curr['amount_eur'], processed_data, 0)

# Output Results
print(f"Processed Records: {processed_data}")
print(f"Total Revenue (EUR): {total_revenue}")
````

## Analysis:

* Modularity: Each step (filtering, transforming, reducing) is isolated.
* Immutability: valid_transactions does not overwrite transactions. processed_data is a new entity.
* Readability: The logic flows linearly from raw data to the final aggregate.
