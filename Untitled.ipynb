# Definite Loop (iteration) "for"

```python
data = [2, 5, 3, 7, 8]
for i, element in enumerate(data):
    if element>5:
        data[i] = 0 #change the element in the data
    print(element)

data
```

```python
data = [2, 5, 3, 7, 8]
N = len(data)
for i in range(N):
    print(i, data[i])
```

# Nested Loop VS Serial Loop

```python
# Nested Loop
for i in range(2):
    for j in range(3):
        print(i,j)
```

```python
# Serial Loop (Sequential Loop)
for i in range(2):
    print(i)
for j in range(3):
    print(j)
```
