## 3. A slow Python loop
We start with normal Python.

````python
def py_square(numbers):
    result = []
    for n in numbers:
        result.append(n*n)
    return result

nums = list(range(200000))
%timeit py_square(nums)
````

## 4. The same code in Cython (no types yet)
Even without types, Cython will make it faster.

````python
%%cython

def cy_square(numbers):
    result = []
    for n in numbers:
        result.append(n*n)
    return result
````

````python
%timeit cy_square(nums)
````

## 5. Adding just **one** typed variable
This makes it even faster.

````python
%%cython

def cy_square_fast(numbers):
    cdef int n
    result = []
    for n in numbers:
        result.append(n*n)
    return result
````

````python
%timeit cy_square_fast(nums)
````

## 6. Simple NumPy example (no advanced features)
Cython improves loops even with NumPy.

````python
import numpy as np

def py_add_one(arr):
    out = np.zeros_like(arr)
    for i in range(len(arr)):
        out[i] = arr[i] + 1
    return out

arr = np.arange(300000)
%timeit py_add_one(arr)
````

````python
%%cython
import numpy as np

def cy_add_one(arr):
    out = np.zeros_like(arr)
    for i in range(len(arr)):
        out[i] = arr[i] + 1
    return out
````

````python
%timeit cy_add_one(arr)
````
