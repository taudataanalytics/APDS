```python
# Importing Python Modules
print("You are not allowed to modify this cell.")

import warnings; warnings.simplefilter('ignore')
import pandas as pd, numpy as np, seaborn as sns
import matplotlib.pyplot as plt
import numba, cython, time
from numba import jit
plt.style.use('bmh'); sns.set()
try:
    seed = int(Student_ID)
    np.random.seed(seed)
except:
    print("Warning! Please write your student number correctly before continuing.")

def generateData(n):
    # Generate synthetic dataset
    # Features: id, group (A/B/C), size, age, rate1, rate2, noise
    ids = np.arange(1, n+1)
    groups = np.random.choice(["A","B","C"], size=n, p=[0.4,0.35,0.25])
    size = np.random.randint(20, 500, size=n)            # numeric
    age = np.random.randint(0, 50, size=n)               # numeric
    rate1 = np.round(np.random.normal(loc=0.5, scale=0.15, size=n), 4)
    rate2 = np.round(np.random.normal(loc=1.0, scale=0.4, size=n), 4)
    # introduce some missing values
    mask = np.random.rand(n) < 0.03
    rate2[mask] = np.nan
    
    df = pd.DataFrame({
        "id": ids,
        "group": groups,
        "size": size,
        "age": age,
        "rate1": rate1,
        "rate2": rate2
    })
    return df.sample(frac=1, random_state=seed).reset_index(drop=True)  # shuffle
```
